## گام ۶: تبدیل‌های شدت (Intensity Transformations)

در این مرحله، مفید است که تصویر را به جای یک آرایه، به عنوان یک تابع $f(x,y)$ در نظر بگیریم، جایی که `x` ایندکس ردیف و `y` ایندکس ستون است. می‌توانید یک تبدیل $T$ را روی تصویر اعمال کنید و یک تصویر جدید بگیرید:
$$
g(x,y)=T(f(x,y))
$$

یک تبدیل شدت فقط به یک نقطه واحد $(x,y)$ بستگی دارد. برای مثال، می‌توانید یک تبدیل خطی $g(x,y) = 2f(x,y) + 1$ اعمال کنید؛ این کار هر پیکسل تصویر را در دو ضرب و یک اضافه می‌کند.

از آنجا که تبدیل‌های شدت فقط به یک مقدار بستگی دارند؛ در نتیجه گاهی اوقات به عنوان نگاشت سطح خاکستری (gray-level mapping) نامیده می‌شوند. اگر $r$ شدت سطح خاکستری باشد، مانند مقادیر هیستوگرام. خروجی جدید $s$ به صورت زیر است:

$$
s=T(r)
$$

</div>


<div dir="rtl">

### نگاتیو تصویر (Image Negatives)

تصویری با $L$ مقدار شدت در محدوده $[0,L-1]$ در نظر بگیرید. می‌توانیم سطوح شدت را با اعمال زیر معکوس کنیم:
$$
g(x,y)=L-1-f(x,y)
$$

با استفاده از نماد تابع تبدیل شدت
$$
s = L - 1 - r
$$

این نگاتیو تصویر (Image Negative) نامیده می‌شود. برای $L=256$ فرمول‌ها ساده می‌شوند به:
$$
g(x,y)=255-f(x,y) \qquad \mbox{و} \qquad s=255-r
$$

</div>
